# BOX PLOTS

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Remove spaces from column names
df.columns = df.columns.str.strip()

# Remove outliers using the IQR method
def remove_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    return data[
        (data[column] >= lower_bound) &
        (data[column] <= upper_bound)
    ]

# Filter outliers
filtered_df = remove_outliers(df, 'Total Fwd Packets')

# Create the box plot
plt.figure(figsize=(12, 8))

sns.boxplot(
    data=filtered_df,
    x='Label',
    y='Total Fwd Packets',
    hue='Label',
    palette='viridis',
    legend=False
)

plt.title('Total Forward Packets by Label', fontsize=16)
plt.xlabel('Label', fontsize=12)
plt.ylabel('Total Forward Packets', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

# PIE CHARTS

```python
import seaborn as sns
import matplotlib.pyplot as plt

# PIE CHART
label_counts = df['Label'].value_counts()

# Calculate percentages
percentages = label_counts / label_counts.sum() * 100

# Filter out labels with less than 5% representation
filtered_labels = percentages[percentages >= 5].index
filtered_counts = label_counts[filtered_labels]

# Create the pie chart
plt.figure(figsize=(8, 8))  # Adjust figure size for better visualization
patches, texts, autotexts = plt.pie(
    filtered_counts,
    labels=filtered_labels,
    autopct='%1.1f%%',
    startangle=90,
    colors=sns.color_palette("viridis", len(filtered_counts)),
    wedgeprops={'edgecolor': 'black', 'linewidth': 0.5}  # Added edge color and line width
)

# Improve aesthetics of the text
for text in texts:
    text.set_fontsize(10)
for autotext in autotexts:
    autotext.set_fontsize(12)
    autotext.set_color('white')

# Add a legend outside the pie chart
plt.legend(patches, filtered_labels, title="Labels", loc="center left", bbox_to_anchor=(1, 0.5))

# Add title and adjust layout
plt.title('Distribution of Labels', fontsize=16)
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.tight_layout()  # Adjust layout for better spacing
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Count estimated protocols
protocol_counts = df['Estimated Protocol'].value_counts()

# Calculate percentages
protocol_percentages = protocol_counts / protocol_counts.sum() * 100

# Keep categories representing at least 5%
filtered_protocols = protocol_percentages[
    protocol_percentages >= 5
].index

filtered_protocol_counts = protocol_counts.loc[filtered_protocols]

# Create pie chart
plt.figure(figsize=(8, 8))

patches, texts, autotexts = plt.pie(
    filtered_protocol_counts,
    labels=filtered_protocol_counts.index,
    autopct='%1.1f%%',
    startangle=90,
    colors=sns.color_palette('magma', len(filtered_protocol_counts)),
    wedgeprops={'edgecolor': 'black', 'linewidth': 0.5}
)

for text in texts:
    text.set_fontsize(10)

for autotext in autotexts:
    autotext.set_fontsize(12)
    autotext.set_color('white')

plt.legend(
    patches,
    filtered_protocol_counts.index,
    title='Protocols',
    loc='center left',
    bbox_to_anchor=(1, 0.5)
)

plt.title('Distribution of Estimated Protocols', fontsize=16)
plt.axis('equal')
plt.tight_layout()
plt.show()
```

# PAIR PLOT

```python

# Faster pair plot with representative samples
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

df.columns = df.columns.str.strip()

selected_features = [
    'Flow Duration',
    'Total Fwd Packets',
    'Total Backward Packets',
    'Active Mean',
    'Idle Mean'
]

required = selected_features + ['Label']

plot_df = df[required].copy()

for col in selected_features:
    plot_df[col] = pd.to_numeric(plot_df[col], errors='coerce')

plot_df = plot_df.replace([float('inf'), -float('inf')], float('nan'))
plot_df = plot_df.dropna(subset=required)

# Sample up to 500 records per label
plot_df = (
    plot_df.groupby('Label', group_keys=False)
    .apply(lambda g: g.sample(n=min(len(g), 500), random_state=42))
    .reset_index(drop=True)
)

print("Rows plotted:", len(plot_df))

g = sns.pairplot(
    data=plot_df,
    vars=selected_features,
    hue='Label',
    diag_kind='hist',  # Much faster than KDE
    plot_kws={'alpha': 0.6, 's': 15},
    diag_kws={'bins': 30},
    height=2.2
)

g.fig.suptitle('Pair Plot of Selected Network Traffic Features', y=1.02)
plt.show()
```

- Distribution of Flow Duration: Most flows have very short durations, as indicated by the steep drop-off in frequency after a short range.

### Pie Chart

The dataset is highly imbalanced, with BENIGN traffic comprising the majority.

- Distribution of Protocols:
Protocol 6 (TCP) dominates the dataset, reflecting its common use in network communication.

## 4. Heatmap and Scatter Plot
### Correlation Heatmap

There’s minimal correlation between Total Backward Packets and other features, suggesting independent behavior.

### Scatter Plot

Outliers with high Flow Duration and Total Forward Packets are likely attack patterns such as DoS or Port Scans.

- Distinct Clustering: BENIGN flows are tightly clustered near the origin, indicating short durations and small packet sizes, which is characteristic of normal traffic.

## 5. Pair Plot
- The relationships between features like Flow Duration, Total Fwd Packets, and Total Backward Packets reveal distinct clusters, potentially aiding in label classification.
- Attack labels such as "DoS Hulk" or "PortScan" tend to show dispersed values, distinguishing them from BENIGN traffic, which is more clustered.
- Feature Relationships: The diagonal histograms reveal that features like Flow Duration and Active Mean are heavily skewed, showing a concentration of values near the lower end
- Label Separation: Certain attack types, such as "DoS Hulk," are distinguishable in specific feature pairings (e.g., Flow Duration vs. Total Fwd Packets), showing potential for effective classification.
